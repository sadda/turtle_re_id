```python
import warnings
from itertools import product

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from tqdm import tqdm
from utils import get_info, plot_accuracy_B, plot_best_scores

from animal_re_id.config import load_paths, resolve_path
from animal_re_id.datasets import SeaTurtleID2022, StripeSpotter, combine_datasets_twe_from_paths
from animal_re_id.models import EncounterPredictions, get_pipeline_aliked, merge_data, visualise_match
from animal_re_id.similarity import SimilarityCache, get_similarity_dataset_config

warnings.filterwarnings("ignore")
```

```python
dataset_name = "TurtlewatchEgypt"
# dataset_name = "SeaTurtleID2022"

ks = [1, 5]
Bs = range(1, 101)
matching_merged = "merged"

B_max = max(Bs)
config = get_similarity_dataset_config(dataset_name, B=B_max)
matching_parts = config.matching_parts
```

```python
local_path = "../../config/paths.json"
paths = load_paths(dataset_name, local_path=local_path)
root_similarity = resolve_path(paths["root_similarity"])
if dataset_name == "TurtlewatchEgypt":
    dataset_true = combine_datasets_twe_from_paths(paths, load_segmentation=False)
    mask1 = dataset_true.metadata["identity"] != "unknown"
    mask2 = ~dataset_true.metadata["identity"].isnull()
    dataset_true = dataset_true.get_subset(mask1 & mask2)
    dataset_true.metadata["year"] = dataset_true.metadata["path"].str[:4].astype(int)
    pipeline_funs_priority = ("miew", 1)
    pipeline_funs_matcher = ("miew-aliked", 1.25)
    y_max1 = 150
    y_max2 = 350
elif dataset_name == "SeaTurtleID2022":
    dataset_true = SeaTurtleID2022.from_paths(paths, False).wildlife_dataset
    pipeline_funs_priority = ("dino", 1)
    pipeline_funs_matcher = ("dino-aliked", 1.0)
    y_max1 = 60
    y_max2 = 300
elif dataset_name == "StripeSpotter":
    dataset_true = StripeSpotter.from_paths(paths, False).wildlife_dataset
    dataset_true = dataset_true.get_subset(range(int(len(dataset_true)/2)))
    pipeline_funs_priority = ("dino", 1)
    pipeline_funs_matcher = ("dino-aliked", 1.0)
    y_max1 = 60
    y_max2 = 300
else:
    raise ValueError("Dataset now known")

split_names = [splitter.get_split_name() for splitter in config.splitters]
pipeline_funs = (pipeline_funs_priority, pipeline_funs_matcher)
identities, n_encounters_known = get_info(dataset_true, config.splitters)
```

```python
k_max = max(ks)
pars_load_all = list(
    product(
        [policy.cache_key for policy in config.ignore_pairs_policies],
        split_names,
    )
)
pars_all = [(B, pipeline_fun, match_opposite_sides, split_name) for match_opposite_sides, split_name in pars_load_all for pipeline_fun in pipeline_funs for B in Bs]
```

```python
cache = SimilarityCache(root_similarity)
img_load = config.img_load

run_specs_by_key = {(rs.pipeline_spec.name, rs.ignore_policy.cache_key, rs.splitter.get_split_name(), rs.matching_part): rs for rs in config.run_specs()}


def restrict_similarity_by_B(similarity_miew, similarity_miew_aliked, B):
    n_rows, n_cols = similarity_miew.shape
    B = min(B, n_cols)
    top_idx = np.argpartition(-similarity_miew, B - 1, axis=1)[:, :B]

    mask = np.ones_like(similarity_miew_aliked, dtype=bool)
    rows = np.arange(n_rows)[:, None]
    mask[rows, top_idx] = False

    similarity = similarity_miew_aliked.copy()
    similarity[mask] = -np.inf
    return similarity


def encounters_to_data(similarity, query, database, matching_part, name, head_multiplier):
    multiplier = head_multiplier if matching_part == "heads" and name != "miew" else 1
    return EncounterPredictions.classify(multiplier * similarity, query, database, matching_part, name, k_max)


name_priority, head_multiplier_priority = pipeline_funs_priority
name_matcher, head_multiplier_matcher = pipeline_funs_matcher

datas = {pars: {} for pars in pars_all}
for match_opposite_sides, split_name in tqdm(pars_load_all):
    for matching_part in matching_parts:
        # Load the saved variables for both pipelines (only at B_max)
        run_spec_priority = run_specs_by_key[(name_priority, match_opposite_sides, split_name, matching_part)]
        saved_vars_priority = run_spec_priority.load(cache)
        similarity_priority = saved_vars_priority["similarity"]

        run_spec_matcher = run_specs_by_key[(name_matcher, match_opposite_sides, split_name, matching_part)]
        saved_vars_matcher = run_spec_matcher.load(cache)
        similarity_matcher_b_max = saved_vars_matcher["similarity"]

        query = saved_vars_matcher["query"]
        database = saved_vars_matcher["database"]

        # Priority ("miew") does not depend on B: compute once and reuse for every B
        data_priority = encounters_to_data(similarity_priority, query, database, matching_part, name_priority, head_multiplier_priority)
        for B in Bs:
            pars = (B, pipeline_funs_priority, match_opposite_sides, split_name)
            datas[pars][matching_part] = data_priority

        # Matcher ("miew-aliked") depends on B: derive B < B_max from the B_max matrices
        for B in Bs:
            similarity = similarity_matcher_b_max if B == B_max else restrict_similarity_by_B(similarity_priority, similarity_matcher_b_max, B)
            data_matcher = encounters_to_data(similarity, query, database, matching_part, name_matcher, head_multiplier_matcher)
            pars = (B, pipeline_funs_matcher, match_opposite_sides, split_name)
            datas[pars][matching_part] = data_matcher
```

```python
parts_all = {}

for pars in tqdm(pars_all):
    data_selection = [datas[pars][m] for m in matching_parts]
    data_merged = merge_data(data_selection)
    datas[pars][matching_merged] = data_merged
```

```python
accuracy_all_list = []
for pars in tqdm(pars_all):
    B, (name, _), match_opposite_sides, split_name = pars

    for matching_part in matching_parts + [matching_merged]:
        data = datas[pars][matching_part]
        for k in ks:
            accuracy = data.compute_matches(k, identities) / n_encounters_known[split_name]
            accuracy_all_list.append({"B": B, "matching_part": matching_part, "name": name, "match_opposite_sides": match_opposite_sides, "split_name": split_name, "k": k, "accuracy": accuracy})
accuracy_all = pd.DataFrame(accuracy_all_list)
```

```python
for k in ks:
    accuracy_avg = accuracy_all[accuracy_all["k"] == k].groupby(["match_opposite_sides", "matching_part", "name", "B"], dropna=False)["accuracy"].mean().unstack("B")

    match_opposite_sides_values = accuracy_avg.index.get_level_values("match_opposite_sides")
    for match_opposite_sides in pd.unique(match_opposite_sides_values):
        mask = match_opposite_sides_values.isna() if pd.isna(match_opposite_sides) else match_opposite_sides_values == match_opposite_sides
        group = accuracy_avg[mask].droplevel("match_opposite_sides")
        plot_accuracy_B(group, matching_parts, figsize=(13, 2), ylabel="accuracy")
        plt.savefig(f"{dataset_name}_accuracy_{k}_{match_opposite_sides}.png", dpi=300, bbox_inches="tight")
```

```python
for k in ks:
    accuracy_avg = accuracy_all[accuracy_all["k"] == k].groupby(["match_opposite_sides", "matching_part", "name", "B"])["accuracy"].mean().unstack("match_opposite_sides")
    if accuracy_avg.shape[1] < 2:
        continue

    accuracy_diff = (accuracy_avg[False] - accuracy_avg[True]).unstack("B")

    plot_accuracy_B(100 * accuracy_diff, matching_parts, figsize=(13, 2), ylim=[-5, 5], ylabel="accuracy gain [%]")
    plt.savefig(f"{dataset_name}_accuracy_{k}_gain.png", dpi=300, bbox_inches="tight")
```

```python
tables = {}
df = accuracy_all[accuracy_all["B"].isin([5, 10, 50, 100])]
for matching_part in df["matching_part"].unique():
    for k in ks:
        sub = df[(df["matching_part"] == matching_part) & (df["k"] == k)]
        table = sub.pivot_table(
            index=["B", "name", "match_opposite_sides"],
            columns="split_name",
            values="accuracy",
            dropna=False,
        ).sort_index()
        table["average"] = table.mean(axis=1)
        tables[(matching_part, k)] = table
```

```python
for k in ks:
    table_average = pd.DataFrame()
    for matching_part in matching_parts + [matching_merged]:
        t = tables[(matching_part, k)].copy()
        t = t.drop(split_names, axis=1).rename({"average": matching_part}, axis=1)
        table_average = pd.concat((table_average, t), axis=1)

    display(table_average)
```

```python
name_desired, _ = pipeline_funs[-1]
best_scores = []
for pars in pars_all:
    B, (name, _), match_opposite_sides, split_name = pars
    if B != max(Bs) or name != name_desired or match_opposite_sides:
        continue

    data = datas[pars]["merged"]
    best_score = data.compute_matches_by_part(1, identities)
    best_scores.extend(best_score)

plot_best_scores(best_scores, matching_parts, y_max=y_max1)
plt.savefig(f"{dataset_name}_merged.png", dpi=300, bbox_inches="tight")
```

```python
for i, (name_desired, _) in enumerate(pipeline_funs):
    best_scores = []
    for pars in pars_all:
        B, (name, _), match_opposite_sides, split_name = pars
        if B != max(Bs) or name != name_desired or match_opposite_sides:
            continue

        for matching_part in matching_parts:
            data = datas[pars][matching_part]
            best_score = data.compute_matches_by_part(1, identities)
            best_scores.extend(best_score)

    use_title = True if i == 0 else False
    plot_best_scores(best_scores, matching_parts, y_max=y_max2, use_title=use_title)
    plt.savefig(f"{dataset_name}_{name_desired}.png", dpi=300, bbox_inches="tight")
```

```python
n_plots = 10
matching_part = "bodies"

name_desired, _ = pipeline_funs[-1]
for pars in pars_all:
    B, (name, _), match_opposite_sides, split_name = pars
    if B != max(Bs) or name != name_desired or match_opposite_sides or split_name != split_names[-1]:
        continue

    data = datas[pars][matching_part]
    break

pipeline = get_pipeline_aliked(cache_prefix=f"{paths['root_features']}/{matching_part}_{config.img_load}")
results = data.extract_highest_predictions(n_plots, n_plots)
for i, result in results.iterrows():
    identity1 = result["identity1"]
    identity2 = result["identity2"]
    if identity1 == identity2:
        visualise_match(data.query, data.database, result, pipeline, "aliked")
        plt.savefig(f"{dataset_name}_{matching_part}_{i:02}_{identity1}.png", dpi=600, bbox_inches="tight")
        plt.close()
```
