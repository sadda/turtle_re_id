```python
import sys
sys.path.append("..")
sys.path.append("../../wildlife-datasets")
sys.path.append("../../wildlife-tools")
import os
import pandas as pd
from wildlife_datasets.datasets import TurtlewatchEgypt_Citizen
from animal_re_id.models import run_detection, turtle_detector
```

```python
root = "/data/wildlife_datasets/turtle_watch/citizen"
file_name = "citizen_files.csv"

data = pd.read_csv(file_name)
transform = lambda data: data[data["year"] == 2026]
```

```python
TurtlewatchEgypt_Citizen.get_data(root, data=data, force=True, transform=transform)
```

```python
dataset = TurtlewatchEgypt_Citizen(root)
dataset.plot_grid();
```

```python
model = turtle_detector()
segmentation = run_detection(dataset, model)
segmentation.to_csv(os.path.join(root, "segmentation.csv"), index=False)
```

```python
dataset = TurtlewatchEgypt_Citizen(root, load_segmentation=True, img_load="bbox")
mask1 = dataset.metadata["label"] == "head"
mask2 = dataset.metadata["score"] >= 0.8
dataset = dataset.get_subset(mask1 & mask2)
dataset.plot_grid();
```
